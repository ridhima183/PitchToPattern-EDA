### Key Insights

- The typical innings score is around **70 runs** at the observed match states.
- The middle 50% of observations fall between **34 and 111 runs**.

# Bivariate Analysis

## What is it?

Bivariate Analysis is the process of studying the relationship between two variables. It helps determine whether changes in one feature are associated with changes in another feature.

## Why is it required?

While univariate analysis examines each feature independently, bivariate analysis explores interactions between features. This helps identify trends, patterns, and relationships that are valuable for prediction and feature selection.

## Questions Answered

- Does one variable influence another?
- Are two variables positively or negatively related?
- Which features are strongly associated with the target variable?
- Do different categories produce different outcomes?

## Relationship between Current Runs and Final Total

### Why analyze this?

The current team score (`runs`) is expected to be one of the strongest indicators of the final innings total.

### Questions Answered

- Does the current score influence the final score?
- Is there a positive relationship?
- Do higher current scores generally lead to higher final totals?

````python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="runs",
    y="total",
    alpha=0.4
)

plt.title("Current Runs vs Final Total")
plt.xlabel("Current Runs")
plt.ylabel("Final Total")

plt.show()
````

````python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df.sample(5000, random_state=42),  # Sample 5,000 points
    x="runs",
    y="total",
    alpha=0.5
)

plt.title("Current Runs vs Final Total")
plt.xlabel("Current Runs")
plt.ylabel("Final Total")

plt.show()
````

### Interpretation

The scatter plot illustrates the relationship between the current team score (`runs`) and the final innings total (`total`). A clear positive relationship is observed, indicating that as the current score increases, the expected final score also tends to increase.

The triangular pattern occurs because the final innings total can never be less than the current score. As the innings progresses and the current score becomes higher, the spread of possible final totals decreases. This indicates that predictions become more reliable in the later stages of the innings.

### Key Insights

- A strong positive relationship exists between current runs and final total.
- The final total is always greater than or equal to the current score.
- Early in the innings, there is greater uncertainty in the final score.
- As the innings progresses, the range of possible final scores narrows.
- The `runs` feature is expected to be one of the strongest predictors of the final innings total.

## Relationship between Overs and Final Total

### Why analyze this?

As the innings progresses, the batting team has more opportunities to score runs. This analysis helps understand how the stage of the innings relates to the final score.

### Questions Answered

- Does the innings stage influence the final score?
- Is there a positive trend?

````python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="overs",
    y="total",
    alpha=0.4,
    color="green"
)

plt.title("Overs vs Final Total")
plt.xlabel("Overs")
plt.ylabel("Final Total")

plt.show()
````

### Interpretation

The scatter plot illustrates the relationship between the innings stage (`overs`) and the final innings total (`total`). Unlike the relationship between current runs and the final score, no strong linear trend is observed. At almost every over, a wide range of final totals is possible, resulting in vertical clusters of observations.

This indicates that the number of overs completed alone is not sufficient to accurately predict the final innings total. The feature becomes more informative when combined with other variables such as the current score, wickets lost, and recent scoring rate.

### Key Insights

- No strong relationship exists between overs and the final total when considered independently.
- A wide range of final scores is observed at every stage of the innings.
- Vertical clusters occur because multiple observations share the same over values.
- The `overs` feature alone has limited predictive power but becomes valuable when used together with other match-related features.

## Relationship between Wickets and Final Total

### Why analyze this?

Losing wickets generally reduces a team's scoring potential. This analysis explores how wickets lost relate to the final innings total.

### Questions Answered

- Do more wickets lead to lower totals?
- Is there a negative relationship?

````python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="wickets",
    y="total",
    alpha=0.5,
    color="red"
)

plt.title("Wickets vs Final Total")
plt.xlabel("Wickets")
plt.ylabel("Final Total")

plt.show()
````

### Interpretation

The scatter plot illustrates the relationship between the number of wickets lost and the final innings total. A moderate negative relationship is observed, indicating that teams losing more wickets generally tend to finish with lower scores.

The vertical clusters occur because wickets are discrete values ranging from 0 to 10. Teams with fewer wickets lost show a much wider range of possible final scores, including many high-scoring innings. As the number of wickets increases, the range of achievable final totals narrows, reflecting the reduced batting resources available.

### Key Insights

- A negative relationship exists between wickets lost and the final innings total.
- Teams with fewer wickets remaining generally have a lower scoring potential.
- High final scores are more common when fewer wickets have been lost.
- The `wickets` feature provides useful predictive information, although its relationship with the target is weaker than that of the `runs` feature.

# How to Read a Correlation Heatmap (Step-by-Step Guide)

A correlation heatmap is one of the most important visualizations in Exploratory Data Analysis (EDA). It helps us understand how numerical variables are related to one another.

Instead of trying to read every number in the heatmap, follow the systematic approach below.

---

# Step 1: Understand What Correlation Means

Correlation measures the strength and direction of the relationship between two numerical variables.

Its value always lies between **-1 and +1**.

| Correlation Value | Meaning | Interpretation |
|-------------------|---------|----------------|
| +1.00 | Perfect Positive | Both variables increase together. |
| +0.70 to +0.99 | Strong Positive | Strong increasing relationship. |
| +0.30 to +0.69 | Moderate Positive | Moderate increasing relationship. |
| 0 | No Linear Relationship | No clear relationship exists. |
| -0.30 to -0.69 | Moderate Negative | One variable increases while the other decreases. |
| -0.70 to -1.00 | Strong Negative | Strong inverse relationship. |
| -1.00 | Perfect Negative | One variable increases exactly as the other decreases. |

---

# Step 2: Ignore the Diagonal

The diagonal always contains **1.00**.

Example:

runs → runs = 1.00

wickets → wickets = 1.00

overs → overs = 1.00

This simply means that every variable is perfectly correlated with itself.

These values do not provide any useful information.

✔ Ignore the diagonal completely.

---

# Step 3: Read Only One Half of the Heatmap

Notice that:

Runs ↔ Wickets = 0.59

Wickets ↔ Runs = 0.59

Both values represent exactly the same relationship.

Therefore, there is no need to read both halves of the heatmap.

Choose either:

- Upper Triangle
or
- Lower Triangle

This immediately reduces the number of values you need to analyze.

---

# Step 4: Start with the Target Variable

Never begin reading the heatmap from the top-left corner.

Instead, identify your **target variable**.

For this project:

Target = total

Now read only the **total row** (or **total column**).

Example:

| Feature | Correlation with Total |
|---------|------------------------|
| runs | 0.25 |
| wickets | -0.34 |
| overs | 0.02 |
| runs_last_5 | 0.37 |
| striker | 0.38 |

Now ask yourself:

**Which features have the strongest relationship with the target?**

Those features are usually the most useful for prediction.

---

# Step 5: Identify Strong Positive Relationships

Now move away from the target variable.

Look for the **largest positive correlation values**.

Example:

Runs ↔ Overs = 0.94

Interpretation:

As the innings progresses, the batting team generally scores more runs.

This relationship makes sense because teams continue accumulating runs throughout the innings.

Another example:

Runs ↔ Runs Last 5 = 0.75

Interpretation:

Teams that have scored many runs recently usually have a higher current score.

Whenever you see a strong positive correlation, ask:

> "Why would these two variables increase together?"

If the answer matches the domain (in this case, cricket), then the relationship is meaningful.

---

# Step 6: Identify Strong Negative Relationships

Now look for blue cells (negative values).

Example:

Wickets ↔ Total = -0.34

Interpretation:

As wickets lost increase, the final innings total generally decreases.

Cricket Explanation:

A team that loses many wickets has fewer batsmen available to continue scoring.

Therefore, higher wickets usually result in lower final totals.

Whenever you see a negative correlation, ask:

> "Why would increasing one variable decrease the other?"

---

# Step 7: Validate Using Real-World Knowledge

This is the most important step.

Never trust numbers blindly.

Always ask:

**"Does this relationship make sense?"**

Example 1

Runs ↔ Overs = 0.94

Does this make sense?

Yes.

Teams score more runs as more overs are played.

---

Example 2

Wickets ↔ Total = -0.34

Does this make sense?

Yes.

More wickets generally reduce a team's scoring ability.

---

Example 3

Overs ↔ Total = 0.02

Does this make sense?

Surprisingly, yes.

Two teams can both be at 10 overs.

Team A:
100/1

Team B:
60/5

Although both teams have completed the same number of overs, their expected final totals are very different.

Therefore, overs alone are not a strong predictor of the final score.

---

# Questions to Ask for Every Correlation

Whenever you look at any correlation value, ask these five questions:

1. Is the relationship positive or negative?

2. Is the relationship strong, moderate, or weak?

3. Does this relationship make logical sense?

4. Can I explain this using a real-world example?

5. How will this relationship help the machine learning model?

---

# Final Checklist for Reading Any Correlation Heatmap

Whenever you see a heatmap, follow this exact order:

✔ Step 1: Understand the correlation scale (-1 to +1)

✔ Step 2: Ignore the diagonal (1.00 values)

✔ Step 3: Read only one half of the heatmap

Following these seven steps allows you to interpret any correlation heatmap confidently and identify the variables that are most useful for analysis and predictive modeling.

# Feature Engineering

## What is Feature Engineering?

Feature Engineering is the process of selecting, modifying, creating, or removing features from a dataset to improve the performance of a machine learning model.

The quality of the input features has a significant impact on the quality of the predictions. Therefore, before training any model, it is important to ensure that only meaningful and relevant information is provided to the algorithm.

---

## Why is Feature Engineering Required?

Real-world datasets often contain features that are:

- Irrelevant to the prediction task.
- Redundant because they provide duplicate information.
- Difficult for machine learning models to interpret.
- Likely to introduce noise or bias.
- Potential sources of data leakage.

Feature engineering improves the dataset by retaining useful information while removing unnecessary complexity.

---

## Questions Answered During Feature Engineering

Before modifying the dataset, we ask the following questions:

- Which features are useful for prediction?
- Which features should be removed?
- Which categorical features need numerical encoding?
- Are there any redundant features?
- Are there any features that could cause data leakage?
- How should the data be prepared before model training?

# Step 1: Identify Irrelevant Features

## What are irrelevant features?

Irrelevant features are variables that do not provide meaningful information for predicting the target variable.

Although these features exist in the dataset, they do not help the machine learning model learn useful patterns.

Removing such features reduces noise and improves model efficiency.

---

## Why do we identify irrelevant features?

Every additional feature increases the complexity of the model.

If a feature contains no useful predictive information, it only increases computation and may reduce model performance.

Therefore, each feature should be evaluated before deciding whether to keep or remove it.

---

# Step 2: Drop Identifier Columns

### Feature

```
mid
```

### What is it?

The `mid` column represents the Match ID.

Each match has a unique identifier.

Example:

```
mid = 1

mid = 2

mid = 3
```

---

### Why remove it?

The Match ID is simply an identifier.

It does not describe any cricket-related information such as runs, wickets or overs.

Machine learning models cannot learn useful scoring patterns from arbitrary IDs.

Keeping this feature only introduces unnecessary information.

---

### Decision

✅ Remove `mid`

# Step 4: Evaluate High Cardinality Features

## What is High Cardinality?

A feature has high cardinality when it contains a very large number of unique categories.

Examples:

- Hundreds of batsmen
- Hundreds of bowlers

---

### Why is this important?

Machine learning models cannot directly understand text.

Categorical variables must first be converted into numerical values.

For features with hundreds of unique categories, encoding creates a very large number of new columns.

This increases:

- Memory usage
- Computational cost
- Model complexity
